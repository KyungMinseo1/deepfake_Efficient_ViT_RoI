```python
from sklearn.model_selection import train_test_split
import pandas as pd
import config, os
from tqdm import tqdm
```

```python
DATA_PATH = os.path.join(config.DATA_DIR, 'archive', 'FaceForensics++_C23')
CSV_PATH = os.path.join(DATA_PATH, 'csv')
tqdm.pandas()
```

```python
ff_metadata = pd.read_csv(os.path.join(CSV_PATH, 'FF++_Metadata_Shuffled.csv'))[['File Path', 'Label']]
ff_metadata['Label_int'] = ff_metadata['Label'].progress_apply(lambda x : 0 if x == 'REAL' else 1)
ff_metadata['data'] = ff_metadata['File Path'].progress_apply(lambda x : os.path.basename(os.path.dirname(x)))
ff_metadata['inner_folder'] = ff_metadata['File Path'].progress_apply(lambda x : os.path.splitext(os.path.basename(x))[0])
ff_metadata['crops_path'] = ff_metadata[['data', 'inner_folder']].progress_apply(lambda x : os.path.join('data', 'archive', 'FaceForensics++_C23', 'crops', x[0], x[1]), axis=1) # type: ignore
```

```python
Y = ff_metadata['Label_int']
X = ff_metadata.drop(['Label_int', 'Label', 'inner_folder'], axis=1)
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, stratify=Y)
```

```python
train_data = pd.concat([x_train, y_train], axis=1).reset_index(drop=True)
test_data = pd.concat([x_test, y_test], axis=1).reset_index(drop=True)
```

```python
train_data.to_csv(os.path.join(CSV_PATH, 'TRAIN.csv'), index=False)
test_data.to_csv(os.path.join(CSV_PATH, 'TEST.csv'), index=False)
```
